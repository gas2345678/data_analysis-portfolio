```python
#Import the required libraries
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
#Import the dataset
df = pd.read_csv("../datasets/london_houses.csv")
```

```python
#We clean our dataset 
df = df.dropna()
```

```python
#Dataset shape
df.shape
```

```python
#Retrieve all the apartements 
apartments = df.loc[df["Property Type"] == "Apartment"]
apartments
```

```python
#All apartments with an area less than 100 square meters
small_apartments = df.loc[(df["Property Type"] == "Apartment") & (df["Square Meters"] < 100)]
small_apartments.head(10)
```

We want to figure out if there is more than houses than apartments by comparing both quantities

```python
#Number of properties for each type
houses_apartments = df.groupby("Property Type")["Property Type"].value_counts()
print(houses_apartments)
```

```python
df = df.rename(columns={"Price (£)": "Price(k)", "Square Meters": "Area"})
```

How much these properties cost in average?

```python
properties_price = df.groupby("Property Type")["Price(k)"].mean()
print(properties_price)
```

Now let's visualize this this information.

```python
import matplotlib.pyplot as plt 
```

```python
plt.bar(properties_price.index,properties_price.values,color=["blue","orange","green","red"])

plt.margins(x=0.1)
plt.margins(y=0.1)
plt.xticks(rotation=45)
plt.xlabel("Property Type")
plt.title("Average Price of properties in London")

plt.ylabel("Average Price (millons)")
plt.show()
```

Detached houses are the more expensive house type in London
