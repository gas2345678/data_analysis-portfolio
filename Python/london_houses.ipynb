# London Houses

This analysis focus on analyzing different houses around London which can have an influence on their respective price like their status, area or number of bedrooms.

A Logistic Regression model can predict the final price based on these features to help house owners to choose the appropiate price for their sales.

```python
#Import the required libraries
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
#Import the dataset
df = pd.read_csv("../datasets/london_houses.csv")
```

```python
#We clean our dataset 
df = df.dropna()
```

```python
#Dataset overview
df.head(10)
```

```python
#Statistical overview of the data
df.describe()
```

```python
#Dataset shape
df.shape
```

```python
#Retrieve all the apartements 
apartments = df.loc[df["Property Type"] == "Apartment"]
apartments
```

```python
#All apartments with an area less than 100 square meters
small_apartments = df.loc[(df["Property Type"] == "Apartment") & (df["Square Meters"] < 100)]
small_apartments.head(10)
```

We want to figure out if there is more than houses than apartments by comparing both quantities

```python
#Number of properties for each type
houses_apartments = df.groupby("Property Type")["Property Type"].value_counts()
print(houses_apartments)
```

As we will work with big quantities, we will replace the pound symbol for the k(thousand) in the "Price" column

```python
df = df.rename(columns={"Price (£)": "Price(k)", "Square Meters": "Area"})
```

How much these properties cost in average?

```python
properties_price = df.groupby("Property Type")["Price(k)"].mean()
print(properties_price)
```

Now let's visualize this this information.

```python
import matplotlib.pyplot as plt 
```

```python
plt.bar(properties_price.index,properties_price.values,color=["blue","orange","green","red"])

plt.margins(x=0.1)
plt.margins(y=0.1)
plt.xticks(rotation=45)
plt.xlabel("Property Type")
plt.title("Average Price of properties in London")

plt.ylabel("Average Price (millons)")
plt.show()
```

Detached house are the most expensive house type in London.

Now we have the most prominent house type, what kind of characteristics can define their price.

```python
detached_houses = df.loc[df["Property Type"] == "Detached House"]
detached_houses
```

```python
print("Total of detached houses: " + str(len(detached_houses)))
```

In general, how big are these houses compared with the other types?

```python
detached_houses_area = detached_houses["Price(k)"].mean()
print(detached_houses_area)
```

Are they new or they were renovated?

```python
detached_houses_status = detached_houses["Building Status"]
print(detached_houses_status.value_counts())
```

```python
plt.bar(detached_houses_status.value_counts().index,detached_houses_status.value_counts().values,color=["blue","orange","green","red"])

plt.title("Building Status of Detached Houses")

plt.xlabel("Building Status")

plt.ylabel("Number of houses")
```
